```python
# Reproduce Metrics (Table 4 equivalent)
scr = sum(r['compliant'] for r in results) / len(results)
vir = sum(1 for r in results if r['new_vulnerabilities'] > 0) / len(results)
print(f'SCR: {scr:.1%} | VIR: {vir:.1%}')
print('📊 Run 5-fold CV locally with `--cv_folds 5` for full statistical results.')
```
